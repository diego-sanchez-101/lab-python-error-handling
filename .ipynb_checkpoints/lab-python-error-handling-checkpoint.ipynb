def initialize_inventory(products):

````python
# Step 1: Define the function for initializing the inventory with error handling
def initialize_inventory(products):
    inventory = {}
    for product in products:
        valid_quantity = False
        while not valid_quantity:
            try:
                quantity = int(input(f"Enter the quantity of {product}s available: "))
                if quantity < 0:
                    raise ValueError("Invalid quantity! Please enter a non-negative value.")
                valid_quantity = True
            except ValueError as error:
                print(f"Error: {error}")
        inventory[product] = quantity
    return inventory

products = ["t-shirt", "mug", "hat", "book", "keychain"]
inventory = initialize_inventory(products)
print(inventory)
````

````python
#Getting customer orders

def get_customer_orders(inventory):
    #Ask for the number of orders until it is valid

    valid_number = False
    while not valid_number:
        try:
            num_orders = int(input("Enter the number of customer orders: "))
            if num_orders <= 0:
                raise ValueError
            valid_number = True
        except ValueError:
            print("Error: Please enter a whole number above 0")
    
    customer_orders = set()
    for i in range(num_orders):
        valid_product = False
        while not valid_product:
            try:
                product = input(f"Enter the name of product {i + 1}: ").strip().lower()
                if product not in inventory:
                    raise ValueError(f"'{product}' is not in the inventory.")
                if inventory[product] <= 0:
                    raise ValueError(f"Sorry, '{product}' is out of stock.")
                valid_product = True
            except ValueError as error:
                print(f"Error: {error} Please try again")
        customer_orders.add(product)
            
    return customer_orders

customer_orders = get_customer_orders(inventory)
print(customer_orders)
````

````python
def calculate_price(customer_orders):
    prices = {}
    for product in customer_orders:
        valid_price = False
        while not valid_price:
            try:
                price = float(input(f"Enter the price of {product} : "))
                if price <= 0:
                    raise ValueError
                valid_price = True
            except ValueError:
                    print("Error: Please enter a whole number above 0")
        prices[product] = price


    total_price = sum(prices.values())
    return (prices, total_price)
    
prices, total_price = calculate_price(customer_orders)
print(prices)
print(f"Total price: {total_price}")
````
